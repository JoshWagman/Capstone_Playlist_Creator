```python
# Create shared input layer from general classifier
shared_input_layer = general_classifier_2.shared_input_layer

# Define MultiTaskModel class (same as in training notebook)
class MultiTaskModel(nn.Module):
    def __init__(self, input_dim, shared_layer=None):
        super(MultiTaskModel, self).__init__()
        # Shared input layer - reference to shared layer object
        if shared_layer is not None:
            self.shared_input_layer = shared_layer
        else:
            self.shared_input_layer = nn.Linear(input_dim, 12)
        
        # User-specific layers
        self.mid_dense_9 = nn.Linear(12, 9)
        self.mid_dense_7 = nn.Linear(9, 7)
        self.output_layer = nn.Linear(7, 3)
        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()
        
    def forward(self, x):
        x = self.relu(self.shared_input_layer(x))
        x = self.relu(self.mid_dense_9(x))
        x = self.relu(self.mid_dense_7(x))
        x = self.sigmoid(self.output_layer(x))
        return x

# Dictionary for models for each user
user_models = {}

# Get the number of users
users = users_data_2['user'].unique()

# Create user models with shared layer and copy weights from general classifier
for user in users:
    # Create model with shared layer
    model = MultiTaskModel(latent_dim, shared_layer=shared_input_layer).to(device)
    
    # Copy weights from general classifier (except shared layer which is already shared)
    model.mid_dense_9.load_state_dict(general_classifier_2.mid_dense_9.state_dict())
    model.mid_dense_7.load_state_dict(general_classifier_2.mid_dense_7.state_dict())
    model.output_layer.load_state_dict(general_classifier_2.output_layer.state_dict())
    
    # Set user-specific layers as trainable (shared layer trainability is controlled separately)
    model.mid_dense_9.requires_grad_(True)
    model.mid_dense_7.requires_grad_(True)
    model.output_layer.requires_grad_(True)
    
    user_models[user] = model

print(f"Created {len(user_models)} user models with shared input layer")
```

```python
# Define step function for PyTorch models (same as in training notebook)
def step(key_order, opt):
    # Accumulate all the operations for the dataset
    preds = {}  # each key will be a separate user
    single_loss = {}
    
    # Set models to training mode
    for x in key_order:
        user_models[x].train()
    
    # Forward pass and backward pass for each user
    for x in key_order:
        opt[x].zero_grad()
        # Make a prediction and calculate loss for each task
        preds[x] = user_models[x](X_train_dict[x])
        loss = criterion(preds[x], y_train_dict[x])
        single_loss[x] = loss.item()
        
        # Backward pass
        loss.backward()
        opt[x].step()
    
    # Calculate overall loss
    y_all = torch.cat([y_train_dict[x] for x in users], dim=0)
    yhat_all = torch.cat([preds[x] for x in users], dim=0)
    single_loss['overall'] = criterion(yhat_all, y_all).item()
    
    return single_loss
```

```python
EPOCHS = 1000

# Need separate optimizers for each output task
opt = {}
for x in users:
    opt[x] = optim.Adam(user_models[x].parameters())

# Loss criterion
criterion = nn.BCELoss()

train_losses = []
test_losses = []

for i in range(EPOCHS):
    
    if i == 750: 
        # Freeze shared layer after 750 epochs
        for x in users:
            user_models[x].shared_input_layer.requires_grad_(False)
            # Recreate optimizers to exclude frozen parameters
            opt[x] = optim.Adam(user_models[x].parameters())

    print('Epoch',i,end=', ')
    random.shuffle(users) # shuffle in place operation, maybe not needed
    
    train_losses.append(step(users,opt)) # batched gradient updates

    y_mtl=[]
    yhat_mtl=[]

    valid_losses = {}
    yhat_test_dict = {}
    for x in users:
        # test on data, for tracking
        user_models[x].eval()
        with torch.no_grad():
            yhat_mlp = user_models[x](X_test_dict[x])
            yhat_test_dict[x] = yhat_mlp
            yhat_mlp_np = yhat_mlp.cpu().numpy()
        y_mtl.append(y_test_dict[x].cpu().numpy().ravel())
        yhat_mtl.append(yhat_mlp_np.ravel())

        # calculate loss
        loss_val = criterion(yhat_mlp, y_test_dict[x])
        valid_losses[x] = loss_val.item()

    
    # Calculate overall validation loss
    y_all_test = torch.cat([y_test_dict[x] for x in users], dim=0)
    yhat_all_test = torch.cat([yhat_test_dict[x] for x in users], dim=0)
    valid_losses['overall'] = criterion(yhat_all_test, y_all_test).item()
    test_losses.append(valid_losses)

    print('Train loss:',np.array(train_losses[-1]['overall']),'Valid loss:',np.array(test_losses[-1]['overall']))
```
